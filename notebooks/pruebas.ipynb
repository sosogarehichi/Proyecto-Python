```python
# filtro por personas que estan entre ocupadas y desocupadas
def filter_by_employment_status(df: pd.DataFrame) -> pd.DataFrame :
    return df[df.ESTADO.isin([1,2])]
```

```python
df_individual["ESTADO"] = df_individual["ESTADO"].astype(int)
```

```python
df_individual_aptos = filter_by_employment_status(df_individual)
```
